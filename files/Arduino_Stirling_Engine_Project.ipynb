# Question 1: Ideal Stirling Engine 

## Part a):

In the following analysis, we will be assuming the gas in the air-tight cylinder in the Stirling enigne is an ideal gas. Making this assumption allows us to utilise the following equation shown below:
                                                       
$$ PV = nRT $$ $$ (1) $$             
The variables present within the equation represent the following quantites:

P - Pressure (Pa)
<br>
V - Volume ($m^3$)
<br>
n - number of gas moles (mol) 
<br>
R - gas constant (J.K/mol)
<br>
T - Temperature (K)
<br>

The following stages present in the stirling engine cycle in terms of the process they undergo are stated below:

State 1 to State 2 - Isothermal Process (No temperature change): $T_1$ = $T_2$ = $T_c$ 
<br>
State 2 to State 3 - Isochoric Process (No volume change):   $V_2$ = $V_3$ 
<br>
State 3 to State 4 - Isothermal Process (No temperature change):  $T_3$ = $T_4$ = $T_h$
<br>
State 4 to State 1 - Isochoric Process (No volume change):  $V_4$ = $V_1$
<br>

Using this informaiton for each process between the states, the ideal gas equations can be manipulated to relate these variables in the different states. We will first write out the relevant ideal gas equations for each state. It is vital to also state these equations below are for a mole number of 1, hence why the variable 'n', is not present:

State 1: 
$$ P_1V_1 = RT_1$$ $$ (2) $$ 

State 2: 
$$ P_2V_2 = RT_2 $$ $$(3) $$ 

State 3: 
$$ P_3V_3 = RT_3 $$ $$(3) $$ 

State 4: 
$$ P_4V_4 = RT_4 $$ $$ (4) $$ 

Process 1 - 2:

The process going from states 1 to 2 is isothermal. This means we can re-arrange the ideal gas equations for state 1 and 2 in order to then equate their respective temepratures to then obtain the relevant equaiton:

$$ P_1V_1/R = P_2V_2/R (5.1) $$

Since 'R' is a constant this can be reduced to:

$$ P_1V_1 = P_2V_2 (5.2) $$


For given values of the pressure at state 1 and volume at state 1 and 2, the pressure at state 2 can be worked out as follows:

$$ P_2 = P_1V_1/V_2 $$ $$ (5.3) $$

Process 2-3:

The process of going form state 2 to state 3 is an isochoric one meaning the volume is the same at both states.

This means we can re-arrange equations 2 and 3 in order to make the volume the subject and equate both equations due to the nature of the process. This would then give the following equation when both are equated:
$$ RT_2/P_2 = RT_3/P_3 $$ $$(6.1) $$

Since 'R' is a constant, this can be cancelled out as well as P_2 made the subject in order to obtain an expression for this. The expression for the pressure at state 3 is as follows:
$$ P_3 = T_3P_2/T_2 $$  $$(6.2) $$

Process 3-4:

The process going from state 3 to 4 is an isothermal process. This is the same as the process of going from state 1 and 2, therefore we can use the same equaiton as $(5.3)$ obtained in order to attain an expression for the pressure at state 4. The change made however would be to use the pressure and volumes at state 3 and 4. The expression is as follows:

$$ P_4 = P_3V_3/V_4 $$ $$ (7.1) $$

Process 4-1:

This is the last process of the cycle and is an isochoric one which is the exact same as the process going from state 2 to 3. Therfore, since this has been derived, we will als be using equation $(6.2)$ however replacing the variables with the relevant ones. The expression yields as follows: 

$$ P_1 = T_1P_4/T_4 $$  $$(6.2) $$

Although when calculating the pressures at each state, this equation is not required as the equations derived before will suffice.

<b>Part i):

In order to derive the equations for the work done by/on the gas in each state of the engine's cycle, it is essential that the original equation for the work done is revisited.

For a gas, the work done can be worked out by calculating the product of the pressure and the change in volume. This can be shown as follows:

$$ W_{i-f} = \int_i^f PdV $$

However, due the pressure being variable during the cycle, the pressure variable has to be replaced by a function of volume representing the pressure. Using the ideal gas equation stated in part a), an equation for pressure in terms of volume can be obtained by re-arranging the equation in order to make the pressure the subject. This will again be carried out assuming the number of moles is 0 (n = 1):

$$ PV = RT $$

Re-arranging the equation to make the pressure the subject leads to:

$$ P = \frac{RT}{V} $$

This can now be subsituted into the work equation derived previously:

$$ W_{i-f} = \int_i^f \frac{RT}{V} dV $$

This can now be analysed for each of the processes occuring between the relevent states. The first process that would be analysed would be from state 1 to 2 in which is an isothermal process. This process being an isothermal process essentially means that the temeprature is a constant and is equal to $ T_c $. This allows the temerpature term to be taken out of the integral:

$$ W_{1-2} = T_c \int_1^2 \frac{R}{V} dV $$

Since $R$ is also a constant (the universal gas constant), this too can be taken out of the integral leaving the following integral:

$$ W_{1-2} = RT_c \int_1^2 \frac{1}{V} dV $$


The integral remaininig is a standrard integral in which the integral of the inverse of a variable results in the $ log_e() $ funciton (also dentoed as $ln()$). Using this fact, this can be applied in order to obtain the final result:

$$ \left[\ln(V)\right]_1^2 $$

Expanding the resulting integral, the following product is obtained:

$$ (ln(V_2)) - (ln(V_1)) $$

To simplifiy this further, the subtraciton logarithm rule can be implemented here which is $log(a) - log(b) = log(\frac{a}{b})$. To conclude the expression for the work done from state 1 to state 2, the constants can $R$ and $T_c$ can be multiplied to the integral to then leave the following equation:

$$W_{1-2} = RT_cln((\frac{V_2}{V_1}) $$

The next process analysed would be from states 2 to 3. However, this process is an isochoric process meaning that there is no volume change as stated before. When observing the equation for the work done by a gas, this consists of a $dV$ term relating to the change in volume. In this case however, this value for this would be 0 based on the nature of the process.

$$ W_{i-f} = \int_i^f PdV $$ 

For this specific process, the work done expression would be the following:

$$ W_{2-3} = \int_2^3 P (0) $$

Which then reduces to the following whereby the integrand becomes 0 leading to the integral value also being 0:

$$ W_{2-3} = \int_2^3 (0) = 0 $$

The work done for this process is therefore 0. This can also be derived via the nature of the graph. The area underneath a pressure - volume graph provides the work done as the process of integration essentially calculates the area under a graph. For the process going from state 2 to 3, we see a vertical line is present which means no area is present underneath this section, again showing that no work is done by the gas during this process. 

The process in deriving the work expression for the process between the states 3 and 4 is exactly the same as between states 1 and 2 due to both being isothermal processes. We can start of by taking out the constant values which in this case would be $R$ and $T_h$ since both $T_3$ and $T_4$ refer to the temperatures at the high temperature thermal resevoir.

$$ W_{3-4} = RT_h \int_3^4 \frac{1}{V} dV $$

The integration can then be carried out as before, leaving the following reuslt:

$$ \left[\ln(V)\right]_3^4 $$

Expanding this out would result in the following:

$$ (ln(V_4)) - (ln(V_3)) $$

Using the logarithm rule stated before as well as multipiyng back the constants taken out before to focus on the integration, the resulting expression for the work done during this process can be obtained:

$$W_{3-4} = RT_hln((\frac{V_4}{V_3}) $$

Since $V_4 = V_1$ and $V_3 = V_2$ due to the isothermal process between the 2 states, the above equations can be expressed as the following :

$$W_{3-4} = RT_hln((\frac{V_1}{V_2}) $$

Finally, the work expression will now be obtained for the process between states 4 and 1. This process is also an isochoric process meaning the volume does not change. This therefore means that the work value automatically becomes 0 as seen with the process between states 2 and 3. This can be represented as follows

$$ W_{4-1} = \int_4^1 (0) = 0 $$





```python
# Part ii):
# Calculating the work done in each process 

W1_2 = (R) * (T_c) * np.log((V2)/(V1))
W2_3 = 0 * 0 # As discussed before as this is an isochoric therefore meaning no work is done by the gas 
W3_4 = (R) * (T_h) * np.log((V4)/(V3))
W4_1 = 0 * 0 # As dicussed before as this is an isochoric therefore meaning no work is done by the gas 

print(W1_2, 'Joules')
print(W2_3,  'Joules')
print(W3_4, 'Joules')
print(W4_1, 'Joules')
```

The values for the work done by/on the gas during each process of the Stirling enigne cycle have been obtained as seen above. The processes occcuring between states 2 and 3 and 4 and 1 consist of no work performed by the gas as discussed beofore.  Negative work done is present with states 1 to 2 as work is done on the system via compression therefore a negative sign is present for convention. The process between states 4 and 1 consists of positive work indicating the system does work in this case. This is shown by expansion of the working fluid which causes this.

## Part B): Flywheel Kinematics

```python
# Part i) 
# The main idea behind calculating revolution times would be to find the time between 2 consecutive trigger values of 0
# This is as this change in trigger value indicates the spoke has cut through the laser beam and so a revolution performed
# An empty list is formed here that will include the times at which a 0 trigger value is obtained from the excel dataset 
zero_times=[]
# The amount of indexes the for loop has to go through should be the same as the length of the trigger column to ensure all trigger values are considered 
values = range(len(trigger))
# The for loop would go through all the trigger values and if a 0 is found, the corresponding time through the use of it's index denoted 'n', is added to the empty list defined above 
for n in values:
    if trigger[n] == 0:
        zero_times.append(time__[n])
# The times are going up in intervals of 19 miliseconds
# To account for the possibility of consecutive zeros which don't mean a revolution, these have to be filtered out
# The np.diff function is used to check the time difference between consecutive values of 0
# If the time difference is 19 miliseconds, these are removed from the final array of revolution times as these don't account for a full revolution 
arr = np.diff(zero_times)  
# Although values of 0.019 or less (which is impossible) have to be accounted, the time differences apart from 0.019 seconds are all larger than 0.08 seconds
# For a better defined graph, the condition of any time values less than 0.08 seconds is used instead
arr[arr < 0.08]=0
# The line above makes time values with a time difference of 0.08 seconds or less as 0
# The line below removes any time difference values with a value of 0 producing a more accurate array of revolution times 
final_array = np.delete(arr, np.where(arr == 0))
display(final_array)
```

```python
# Part ii):
# The number of values in the list outputted before that provides the time for the revolution would provide us with the amount of revolutions 
# We shall use the length funciton again to allow us to do this 
display(len(final_array))
```

```python
# Part iii):
# We will now be calculating the instantaneous angular velocity
# This can be done via dividing the angle rotated by the engine wheel per revolution which is 2 pi radians by the time taken for each revolution 
# The for loop allows would mean that every element in the array with the time of revolutions are used individually to then carry out the calculation 
angular_velocity = []
constant = 2* (np.pi)
for x in final_array:
    angular_velocity.append(constant/x)
display(angular_velocity)
```

```python
# Part iv):
# A graph of the angular velocity against the time for each revolution will be plotted 
plt.plot(angular_velocity, color='m')
plt.title('Angular velocity against the revolution number')
plt.xlabel('Revolution number (no units)')
plt.ylabel('Angular velocity (Rad$s^-1$)')
```

```python
# Part v):
# The values for the angluar velocity obtained previously in radians per second will now be converted into revolutions per minute
angular_velocity_rpm = []
# The new constant would account for divding by 2 pi radians to obtain the number of revolutions
# It would also account for multiplying by 60 to account for the number of revolutions in a minute 
constant_new = (60) / (2* (np.pi))
for p in angular_velocity:
    angular_velocity_rpm.append(constant_new*p)

# A plot of the angular velocity against revolution number will be plotted here 
plt.plot(angular_velocity_rpm, color='c')
plt.title('Angular velocity against the revolution number')
plt.xlabel('Revolution number (no units)')
plt.ylabel('Angular velocity (RPM)')
```

<b>Part 2b vi):

Decreasing the length of the spoke by a half would cause for the angular velocity value of the Stirling engine to change. This can be exlpained via the conservation of angular momentum. The initial moment of inertia term would be greater than if the spoke length was to be halved due to more material present away from the axis of rotation. Since no angular impulse acts, and only the rotation of the Stirling engine wheel about its axis occurs, the product of the moment of inertia and angular velocity for a wheel consisting of the original spoke length would equal that of teh product consisting of spokes with half the length. Via ratio analysis, it can be deduced that since the moment of inertia term is smaller when the wheel consists of smaller spokes, a greater angluar velocity term has to therefore account for this meaning an increase in angular velcoity with a decrease in spoke length. The explanation can be presented mathemtcailly as follows:

$$ I_{G} = \int r^2 dm $$

This shows that with more material away from the centre of axis (greater spoke length) the moment of inertia is greater.
Presenting the angular momentum conservation equaiton for this case:
$$ I_Gw_1 = I_gw_2$$  

In this case, $ I_G $ represents the moment of inertia before and $ I_g $ after. Since $ I_G > I_g  $ this therefore means that $w_2 > w_1$ via ratios, hence an increase in angular velocity with a decrease in spoke length.




<b>Part 2b vii):

When considering the effect of an increase in the width of the spokes(via doubling) on the angular velocity, the same explanation for the previous scenario can also be used here. The mass moment of inertia with the expression shown previosuly, is directly proprtional to the mass of the object as well as distance from the centre of the axis. In this case, when doubling the  width of the spokes, this would cause a decrease in the engine's angular velocity. This is also since the angular momentum is conserved whereby no external impulse is applied.

$$ I_{G} = \int r^2 dm $$

$$ I_Gw_1 = I_gw_2$$  

$ I_G $ - Mass moment of inertia before
<br>
$ I_g $ - Mass moment of inertia after 

Since $ I_g $ > $ I_G $, when the width of the spokes is doubled, this would mean that $ w_2 $ < $ w_1 $. 

As a result of the angular velocity decreasing, this would also mean that the RPM values also decrease since both variables are directly proprtional.

3. The phase angle of the Stirling engine can be optimized depending on if it is running on a higher or low angular velocity. On a higher RPM the phase angle can be greater than 90 degrees proving more time for the air to warm up and cool down. Similarily for lower RPM values, this should be lower than 90 dgerees in order to provide less time for the air in the engine to warm up and cool down resulting in a greater power output. Another change that can be made is to minimze the dead space volume after compression as much as possible. This is as the hot air within the Stirling enigne would mix with the air in the dead volume therfore decrease the temperature gradient between the 2 surfaces and so decrease the efficiency value which is not desirable.
